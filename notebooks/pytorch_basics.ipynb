```python
import torch

x = torch.tensor([[1.0, 2.0], [3.0, 4.0]])
print(x.shape)
print(x.dtype)
print(x + 1)
print(x @ x)
    
print(x.mean())
print(x.sum(dim = 0))
```

```python
w = torch.tensor(0.3, requires_grad = True)

y = w ** 2

y.backward()

print(y)
print("Градиент (производная):", w.grad)
```
